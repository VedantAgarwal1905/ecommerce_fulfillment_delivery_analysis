```python
df_orders['delivery_time_days'] = ((df_orders['order_delivered_customer_date'] -df_orders['order_delivered_carrier_date']).dt.total_seconds() / (24 * 3600)).round(2)
```

```python
df_orders['delivery_delay_days'] = ((df_orders['order_delivered_customer_date'] -df_orders['order_estimated_delivery_date']).dt.total_seconds() / (24 * 3600)).round(2)
```
